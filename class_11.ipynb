```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'category': ['A', 'B', 'A', 'C', 'B'],
    'value': [10, 20, 15, 30, 35]
})

df['category'] = df['category'].astype('category')
print(df.dtypes)
```

Vectorize operation

```python
df['new_value'] = df['value']*2
```

Slower

```python
df['new_value'] = df.apply(lambda raw: raw['value']*2, axis=1)
```

Using eval() and query()

```python
df.eval('new_value = value*2', inplace = True) # faster then apply
df_filtered = df.query('value > 20') #Efficiency filterning
```

MultiIndex for Hierarchical Data

```python
array = [
    ['A', 'A', 'B', 'B'],
    [1, 2, 1, 2]
]
index = pd.MultiIndex.from_arrays(array, names=('Group', 'Number'))
df = pd.DataFrame({'Value': [10, 20, 15, 30]}, index=index)
print(df)
```

```python
array = [
    ['A', 'A', 'B', 'B'],
    [1, 2, 1, 2]
]
index = pd.MultiIndex.from_arrays(array, names=('Group', 'Number'))
df = pd.DataFrame({'Value': [10, 20, 15, 30]}, index=index)
print(df)
print(df.loc['A', 1])
```

Pivot Table

```python
import pandas as pd
df = pd.DataFrame({
    'Category': ['A', 'A', 'B', 'B'],
    'Type' : ['X', 'Y', 'X', 'Y'],
    'Value': [10, 20, 15, 25]

})
pivot_df = df.pivot_table(values='Value', index='Category', columns='Type', aggfunc='sum')
print(pivot_df)
```

```python
df['rolling_avg'] = df['Value'].rolling(window=2).mean()
print(df)
```

```python
import sqlite3
import pandas as pd

# Connect to the database (or create it if it doesn't exist)
conn = sqlite3.connect('mydatabase.db')

# Create a cursor object to execute SQL commands
cursor = conn.cursor()

# Create the table if it doesn't exist
cursor.execute('''
CREATE TABLE IF NOT EXISTS mytable (
    column1 TEXT,
    column2 INTEGER,
    -- Add other columns as needed
)
''')

# Commit the changes to the database
conn.commit()

# Now you can read data from the table
df = pd.read_sql_query("SELECT * FROM mytable", conn)
print(df.head())

# Close the connection when done
conn.close()
```
