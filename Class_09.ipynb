Element-wise Operation:

```python
import numpy as np
a = np.array([1, 2, 3])
b = np.array([4, 3, 2])
print(a + b)
print(a - b)
print(a * b)
print(a / b)
```

Aggregate Functions:

```python
array = np.array([1, 2, 3, 4, 5])
print(np.sum(array))
print(np.mean(array))
print(np.min(array))
print(np.max(array))
print(np.std(array))
print(np.var(array))
```

Matrix Multiplication:
(1.)This line calculates the standard deviation of the elements in the array using the np.std() function and prints the result. Standard deviation is a measure of the spread or dispersion of data around the mean.
(2.)This line calculates the variance of the elements in the array using the np.var() function and prints the result. Variance is another measure of the spread of data, and it is the square of the standard deviation.

```python
a = np.array([[1, 2], [3, 4]])
b = np.array([[5, 6], [7, 8]])
print(np.dot(a, b))
print("\n")
print(np.matmul(a, b))
print("\n")
print(np.dot(b, a))
print("\n")
print(np.matmul(b, a))
```

**Broadcasting:**

```python
a = np.array([1, 2, 3])
b = 2
print(a + b)
print("\n")
print(a - b)
print("\n")
print(a * b)
print("\n")
print(a / b)
```

**Aplication in Data Science and Machine Learning**

Data Preprocessing:

```python
data = np.array([1, 2, 3, 4, 5])
normalized_data = (data - np.min(data)) / (np.max(data) - np.min(data))
print(normalized_data)
```

Random Sampling:

```python
random_data = np.random.rand(5)
print(random_data)
```

Vectorized Operation:

```python
a = np.array([1, 2, 3, 4])
b = np.array([4, 5, 6, 7])
c = np.array([7, 8, 9, 1])
result1 = a*b
result2 = a * b + c
print(result1)
print("\n")
print(result2)
```

**Advanced Topic and Optimization in NumPy**

Memory Management and Views:

```python
a = np.array([1, 2, 3, 4, 5])
b = a[2:4]
b[0] = 100
print(a)
```

In-place Operation:

```python
a = np.array([1, 2, 3, 4, 5])
a*= 2
print(a)
```

Optimized Linear Algebra:

```python
a = np.array([[1, 2], [3, 4]])
b = np.array([[5, 6], [7, 8]])

print(np.dot(a, b))
print("\n")

```
